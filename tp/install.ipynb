# Installation de l'environnement Julia pour les TPs

Ce notebook crée l'environnement Julia utilisé par les TPs. Avant de l'exécuter, installez Julia, Git et VSCode, puis clonez le dépôt du cours en suivant les [instructions du README](https://github.com/ocourses/calcul-differentiel-edo/blob/main/README.md). Exécutez ensuite les cellules dans l'ordre, une première fois après l'installation de Julia.

L'installation peut prendre plusieurs minutes la première fois : Julia doit télécharger et précompiler les packages et leurs dépendances. Les exécutions suivantes seront généralement plus rapides.

Deux fichiers seront créés dans le répertoire `tp/` :

- `Project.toml` décrit les packages utilisés directement par les TPs ;
- `Manifest.toml` enregistre les versions exactes de ces packages et de leurs dépendances, afin de retrouver le même environnement d'une machine à l'autre.

Le fichier `Project.toml` peut évoluer lorsque de nouveaux TPs nécessitent de nouveaux packages. Le fichier `Manifest.toml` sera alors mis à jour automatiquement.

Après cette installation, chaque notebook doit activer cet environnement avant de charger les packages. Les notebooks du cours contiennent déjà cette activation. Dans une session Julia interactive, utilisez :

```julia
using Pkg
Pkg.activate(@__DIR__)
```

Dans un notebook situé dans un sous-répertoire de `tp/`, utilisez le chemin vers le répertoire parent, par exemple `Pkg.activate(joinpath(@__DIR__, ".."))`.

````julia
# activate the project in the directory containing this notebook
using Pkg
Pkg.activate(@__DIR__)
````

````julia
# install or update the packages in the local project
# This cell can take several minutes the first time.
Pkg.add(["DualNumbers", 
         "ForwardDiff", 
         "LinearAlgebra", 
         "OrdinaryDiffEq", 
         "Plots", 
         "Polynomials", 
         "Printf"])

# resolve and install all dependencies recorded in the environment
Pkg.instantiate()
````

````julia
# verify that the environment is active and that all required packages load correctly
Pkg.status()

using DualNumbers
using ForwardDiff
using LinearAlgebra
using OrdinaryDiffEq
using Plots
using Polynomials
using Printf

println("Installation réussie : tous les packages des TPs sont disponibles.")
````
