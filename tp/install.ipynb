# Installation de l'environnement Julia pour les TPs

Ce notebook installe l'environnement Julia utilisé par les TPs. Avant de l'exécuter, installez Julia, Git et VSCode, puis clonez le dépôt du cours en suivant les [instructions du README](https://github.com/ocourses/calcul-differentiel-edo/blob/main/README.md).

## Préparer l'exécution des notebooks

Un notebook Julia a besoin du paquet `IJulia` pour démarrer son noyau. Installez-le une première fois depuis un terminal, avec la même version de Julia que celle que vous utiliserez dans VSCode :

```bash
julia -e 'using Pkg; Pkg.add("IJulia"); using IJulia; installkernel("Julia")'
```

Cette commande installe `IJulia` dans l'environnement général de Julia et enregistre le noyau Julia pour Jupyter. Elle doit être exécutée dans un terminal, et non dans une cellule : le noyau doit être démarré avant que les cellules du notebook puissent être exécutées.

Dans VSCode, ouvrez ensuite ce notebook et choisissez le noyau Julia installé (commande **Select Kernel** en haut à droite, puis **Julia**). Si plusieurs versions de Julia sont installées, choisissez celle avec laquelle `IJulia` a été installé.

## Installer les packages des TPs

L'environnement du cours est décrit par deux fichiers **déjà présents** dans le répertoire `tp/` :

- `Project.toml` — la liste des packages utilisés directement par les TPs ;
- `Manifest.toml` — les versions **exactes** de ces packages et de toutes leurs dépendances, afin d'obtenir le même environnement d'une machine à l'autre.

Il n'y a donc rien à ajouter : les cellules ci-dessous activent cet environnement et installent les versions enregistrées dans `Manifest.toml`. La première exécution peut prendre plusieurs minutes (téléchargement et précompilation).

Ne supprimez pas ces deux fichiers et ne les remplacez pas par ceux d'un autre projet Julia. Ils peuvent évoluer lorsqu'un nouveau TP nécessite un package : récupérez alors la mise à jour avec `git pull`, puis réexécutez ce notebook.

Chaque notebook de TP active lui-même cet environnement (`Pkg.activate(@__DIR__)`) avant de charger ses packages ; vous n'avez pas à le refaire. Dans un notebook situé dans un sous-répertoire de `tp/`, l'activation utilise le chemin du répertoire parent, par exemple `Pkg.activate(joinpath(@__DIR__, ".."))`.
